# Data Types and data structure

## List

```python
# In[1]:
my_list = [10, "apple", 3.14, True, 10]

print(f"List: {my_list}")
print(f"Type: {type(my_list)}")

# Example of mutability (changing an element)
my_list[1] = "banana"
print(f"List after modification: {my_list}")
```

# Tuple

```python
# In[2]:
my_tuple = (100, 'orange', False)

print(f"Tuple: {my_tuple}")
print(f"Type: {type(my_tuple)}")
```

# Set

```python
# In[3]:
# Note: The duplicate '2' is automatically removed
my_set = {1, 2, 3, 2, 4, 1}

print(f"Set: {my_set}")
print(f"Type: {type(my_set)}")

# Example of mutability (adding an element)
my_set.add(5)
print(f"Set after adding '5': {my_set}")
```

# Dictionary

```python
# In[4]:
my_dictionary = {
    "name": "Gemini",
    "version": 2.5,
    "framework": "Python"
}

print(f"Dictionary: {my_dictionary}")
print(f"Type: {type(my_dictionary)}")

# Accessing a value by its key
print(f"Model Name: {my_dictionary['name']}")

# Example of mutability (updating a value)
my_dictionary["version"] = 3.0
print(f"Updated Dictionary: {my_dictionary}")
```

# String

```python
# In[5]:
my_string = "Hello, Python World!"

print(f"String: {my_string}")
print(f"Type: {type(my_string)}")

# Accessing a character
print(f"First character: {my_string[0]}")

# Length of the string
print(f"Length: {len(my_string)}")
```

# Number

```python
# In[6]:
my_integer = 42
my_float = 98.6

print(f"Integer: {my_integer} (Type: {type(my_integer)})")
print(f"Float: {my_float} (Type: {type(my_float)})")

# Simple operation
result = my_integer / 5
print(f"Division result: {result} (Type: {type(result)})")
```

# Boolean

```python
# In[7]:
is_active = True
is_logged_in = False

print(f"Active Status: {is_active} (Type: {type(is_active)})")
print(f"Logged In Status: {is_logged_in} (Type: {type(is_logged_in)})")

# Comparison resulting in a boolean
comparison = (5 > 3)
print(f"Is 5 > 3? {comparison}")
```
